# Wrap-Up
> By now, you’ve seen how probability gives us a powerful framework for describing uncertainty, making predictions, and reasoning with incomplete information.

Let’s briefly recap what you’ve learned:

## Core Ideas We Built Up
- Counting principles — how to rigorously determine the size of event spaces
- Permutations and combinations — the foundation for understanding likelihoods
- Probability fundamentals — how to define probabilities and use axioms
- Conditional probability — how to update beliefs when new information arrives
- Bayes’ Theorem — the formal link between prior beliefs and new evidence
- Independence — when events don’t affect each other, and how to prove it

## From Probability to Statistical Modeling
Most of this course focused on single events or variables and the maths behind probabilities, but these ideas extend naturally to modeling the relationships between variables — which is the heart of statistics.

For example:

### (1) In regression, we’re estimating the conditional distribution of one variable given another:

$P(Y∣X)$

### (2) In classification, we often want to know:

What is the probability that an observation belongs to class $A$ given its features?

### (3) Even advanced machine learning models are built on chains of conditional probabilities.

## A Unifying Perspective

The underlying question in statistics is:

Given what I know (data), what can I infer (or predict)?

Probability theory gives us a language to answer that — whether we do so as frequentists or Bayesians.

In this course, you've not only learned how to calculate probabilities — I hope you've developed the conceptual tools to interpret, challenge, and build statistical models with greater confidence.
