# 7 Independence

Independence is a fundamental concept in probability theory. It is a measure of the relationship between two events. Two events are independent if the occurrence of one event does not affect the occurrence of the other event.



:::{admonition} Mutually Exclusivity
:class: definition

Two events $A$ and $B$ are **mutually exclusive** if they cannot occure at the same time. In other words, if $A$ occurs, then $B$ cannot occur and vice versa.

:::

If we calculate the probability $P(A \cup B)$ of two mutually exclusive events, we can write:

$$
P(A \cup B) = P(A) + P(B)
$$

In general, the probability of the union of two events is the sum of the probabilities of the events minus the probability of the intersection of the two events:

$$
P(A \cup B) = P(A) + P(B) - P(A \cap B)
$$

But, if the events are mutually exclusive, then the probability of the intersection of the two events is 0.

:::{admonition} Independence
:class: definition

Two events $A$ and $B$ are **independent** if the occurrence of one event does not affect the occurrence of the other event. In other words:

$$
P(A|B) = P(A)
$$

:::

:::{admonition} Probability of the Intersection of two Independent Events
:class: rule

If two events are independent, then the probability of the intersection of the two events is the product of the probabilities of the events:

$$
P(AB) = P(A)P(B)
$$

:::

:::{admonition} Probability of the Intersection of two Independent Events
:class: proof

$$
P(AB) = P(A|B)P(B) \underbrace{=}_{P(A|B) = P(A)} P(A)P(B)
$$

:::


:::{admonition} If A is independent of B, then B is independent of A
:class: proof

$$
P(A|B) = P(A) \implies P(B|A) = \frac{P(AB)}{P(A)} = \frac{P(A)P(B)}{P(A)} = P(B)
$$

:::

:::{admonition} If $A$ is independent of $B$, then $A$ is independent of $B^c$
:class: proof

$$
\begin{align*}
P(AB^c) &= P(A) - P(AB) & \text{(by the law of total probability)} \\
&= P(A) - P(A)P(B) & \text{(by independence)} \\
&= P(A)(1 - P(B)) & \\
&= P(A)P(B^c) & \text{(identity axiom)}
\end{align*}
$$
